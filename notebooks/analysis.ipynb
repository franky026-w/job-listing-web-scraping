# Automated Job Listing Collection for Recruiters

## Business Problem
Recruiters and career service teams often need to review job postings from websites and organize them into a structured format. Doing this manually takes time and may lead to data entry errors.

## Automation Goal
This Python automation collects job posting information from a website and exports the results into a CSV file.

## Input
- A webpage containing job listings

## Output
- A CSV file with job title, company name, and location

## Business Benefit
This automation saves time, improves efficiency, reduces manual work, and creates a structured file that can be used for further reporting or analysis.

```python
import requests
from bs4 import BeautifulSoup
import csv
from google.colab import files
```

```python
# Request the webpage
url = "https://realpython.github.io/fake-jobs/"
page = requests.get(url)

# Check if the request was successful
print("Status code:", page.status_code)
```

```python
# Parse the HTML content
soup = BeautifulSoup(page.content, "html.parser")

# Find all job cards
jobs = soup.find_all("div", class_="card-content")

print("Number of job listings found:", len(jobs))
```

```python
# Create and write to a CSV file
with open("jobs.csv", mode="w", newline="", encoding="utf-8") as csv_file:
    writer = csv.writer(csv_file)

    # Write the header row
    writer.writerow(["Title", "Company", "Location"])

    # Loop through each job posting
    for job in jobs:
        title = job.find("h2", class_="title").text.strip()
        company = job.find("h3", class_="company").text.strip()
        location = job.find("p", class_="location").text.strip()

        writer.writerow([title, company, location])

print("Job data has been saved to jobs.csv")
```

```python
# Read the CSV file and preview the first few lines
with open("jobs.csv", "r", encoding="utf-8") as file:
    for i, line in enumerate(file):
        print(line.strip())
        if i >= 5:
            break
```
